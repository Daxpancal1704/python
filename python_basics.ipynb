## 1. Basic Input / Output

- `print()` sends output to the console.
- `input()` reads a line of text from the user (always returns a **string**).
- Use f-strings (`f"..."`) for clean, readable formatting — this is the modern standard.


```python
# Basic print
print("Hello, world!")

# Printing multiple values
name = "Riya"
age = 22
print("Name:", name, "Age:", age)

# f-strings (preferred way to format output)
print(f"{name} is {age} years old.")

# print() extra arguments: sep and end
print("A", "B", "C", sep=" | ", end=" <-- end\n")
```

```python
# input() always returns a string — you must convert it if you need a number
user_name = input("Enter your name: ")
print(f"Hello, {user_name}!")

# Example of type conversion
age_str = input("Enter your age: ")
age = int(age_str)
print(f"Next year you will be {age + 1}")
```

### Key Points (Basic I/O)
- `input()` **always** returns a `str`. Forgetting to cast (`int()`, `float()`) is one of the most common beginner bugs (e.g. `"5" + "3"` gives `"53"`, not `8`).
- f-strings support expressions and formatting: `f"{price:.2f}"`, `f"{value:,}"`, `f"{x!r}"`.
- `print()` has useful but often-ignored parameters: `sep`, `end`, `flush`.


## 2. Lists, Tuples, Sets, Dictionaries

| Structure | Ordered | Mutable | Duplicates | Syntax |
|---|---|---|---|---|
| List | Yes | Yes | Yes | `[1, 2, 3]` |
| Tuple | Yes | No | Yes | `(1, 2, 3)` |
| Set | No | Yes | No | `{1, 2, 3}` |
| Dictionary | Yes (3.7+) | Yes | Keys: No | `{"a": 1}` |


### List 
- ordered : Items maintain their exact insertion order.
- mutable : You can modify, add, or remove items after creation.
- dynamic :  They automatically grow or shrink in size as needed.
- heterogeneous : A single list can hold integers, strings, booleans, or other lists

```python
# ---- LIST ----
fruits = ["apple", "banana", "cherry"]
fruits.append("mango")          # add to end
fruits.insert(1, "kiwi")        # insert at index
fruits.remove("banana")         # remove by value
print(fruits)
print("Length:", len(fruits))
print("Slice [1:3]:", fruits[1:3])
print("Reversed:", fruits[::-1])

# List comprehension - offers a shorter, cleaner syntax to create a new list
# new_list = [expression for item in iterable if condition]
squares = [x**2 for x in range(1, 6)]
print("Squares:", squares)
```

```python
# ---- TUPLE ----
point = (10, 20)
print("Point:", point)
x, y = point   # unpacking
print(f"x={x}, y={y}")

# Tuples are immutable
try:
    point[0] = 99
except TypeError as e:
    print("Error:", e)
```

### copy() 
- shallow copy -> list.copy(), copy.copy(), list[:]
- deep copy -> deepcopy()

```python
# A shallow copy creates a new outer object, but the elements inside are still shared.
list1 = [10, 20, 30]
list2 = list1.copy()

list2.append(40)
print(list1)
print(list2)
```

```python
list1 = [
    [1,2],
    [3,4]
]
list2 = list1.copy()

list2[0].append(100)

print(list1)
print(list2)
#Only the outer list is copied. point to the same inner list
```

```python
import copy

list1 = [[1, 2], [3, 4]]
list2 = copy.copy(list1)

print(list1 is list2)         # False (different outer lists)
print(list1[0] is list2[0])   # True (same inner list)
```

```python
# A deep copy creates completely independent copies of every nested object.
import copy

list1 = [
    [1,2],
    [3,4]
]

list2 = copy.deepcopy(list1)

list2[0].append(100)

print(list1)
print(list2) 
```

```python
# ---- SET ----
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference (a-b):", a - b)
print("Symmetric difference:", a ^ b)

a.add(10)
a.discard(1)
print("Updated set:", a)

# Sets automatically remove duplicates -- handy for de-duplication
nums = [1, 2, 2, 3, 3, 3, 4]
print("Unique values:", set(nums))
```

### Frozen set 

It stores only unique values and does not allow modification after creation. It is immutable

```python
a = ()
f1 = frozenset(a)
print(f1)    # frozenset(a) creates an empty frozenset.

b = ["Geeks", "for", "Geeks"]
f2 = frozenset(b)  # frozenset(b) removes duplicate "Geeks".
print(f2)

d = {"name": "A", "age": 21}
f = frozenset(d)
print(f)   #  frozenset(d) stores only dictionary keys. Values are not included in the frozenset.
```

```python
# ---- DICTIONARY ----
student = {
    "name": "ram",
    "age": 21,
    "skills": ["Python", "SQL"]
}

print(student, end = "\n \n")
print(student["name"])
print(student.get("grade", "Not Available"))  # Try to get the value associated with key. If the key doesn't exist, return default_value instead of giving an error.

student["age"] = 22          # update
student["grade"] = "A"       # add new key
del student["skills"]        # remove key
print(student)

for key, value in student.items():
    print(f"{key}: {value}")

# Dictionary comprehension
squares_dict = {x: x**2 for x in range(1, 6)}
print(squares_dict)
```

## 3. String Manipulation

Strings are immutable sequences of characters. Every "modification" method returns a **new** string.


```python
s = "  Hello, Python World!  "

print(s.strip())               # remove leading/trailing whitespace
print(s.strip().lower())       # lowercase
print(s.strip().upper())       # uppercase
print(s.strip().replace("Python", "Coding"))
print(s.strip().split(","))    # split into a list

words = ["Learn", "Python", "Fast"]
print(" ".join(words))         # join list into string

# Indexing and slicing
text = "Internship"
print(text[0], text[-1])       # first, last character
print(text[0:4])               # slicing
print(text[::-1])              # reverse a string

# Useful checks
print(text.startswith("Inter"))
print(text.endswith("ship"))
print(text.isdigit())

# f-string formatting
pi = 3.14159265
print(f"Pi rounded: {pi:.2f}")
print(f"Padded: {5:03d}")


```

## 4. Operators

Arithmetic, comparison, logical, assignment, membership, and identity operators.


```python
# Arithmetic operators
a, b = 17, 5
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)        # always returns float
print("Floor Division:", a // b) # rounds down
print("Modulus:", a % b)         # remainder
print("Exponent:", a ** b)
```

```python
# Comparison operators -> always return a boolean
# print(a == b, a != b, a > b, a < b, a >= b, a <= b)

# Logical operators
x, y = True, False
print("AND:", x and y)
print("OR:", x or y)
print("NOT:", not x)

# Short-circuit example
def noisy(val, label):
    print(f"Evaluating {label}")
    return val

result = noisy(False, "left") and noisy(True, "right")  # right never runs
print("Result: ", result)
print("\n")

result2 = noisy(False, "left") or noisy(True, "right")  # this time it will check right side also 
print("Result 2 :", result2)
```

```python
# Assignment operators
n = 10
n += 5   # n = n + 5
n -= 2
n *= 3
print("n after compound assignment:", n)

# Membership operators
fruits = ["apple", "banana", "mango"]
print("apple" in fruits)
print("grape" not in fruits)

# Identity operators (is / is not) -- compare object identity, not value
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print(list1 == list2)   # True -- same VALUES
print(list1 is list2)   # False -- different OBJECTS in memory

print(id(list1))
print(id(list2))


print([] == [])
print([] is [])

a = None
b = None 
print(a == b)
print(a is b) # There is only one None object likewise for true and false also 


a = 256
b = 256

print(a == b)
print(a is b)  # (typically -5 to 256) for small integer it will print true 

a = 1000
b = 1000

print(a == b)
print(a is b)
```

## 5. Conditional Statements

`if`, `elif`, `else` control the flow of a program based on boolean conditions.


```python
marks = 78

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Grade: {grade}")

# Nested conditionals
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Entry denied")

# Ternary (conditional expression) -- compact one-line if/else
status = "Adult" if age >= 18 else "Minor"
print(status)
```

```python
number = 2

match number:
    case 1:
        print("One")
    case 2 | 3:
        print("Two or Three")
    case _:
        print("Other number")
```

## 6. Loops

`for` loops iterate over a sequence; `while` loops repeat while a condition is true.


```python
# for loop over a range
for i in range(1, 6):
    print(i, end=" ")
print()

# for loop over a list
colors = ["red", "green", "blue"]
for color in colors:
    print(color.upper())

# enumerate -- get index + value together
for index, color in enumerate(colors):
    print(index, color)

# looping over a dictionary
scores = {"Alice": 85, "Bob": 92}
for name, score in scores.items():
    print(f"{name}: {score}")


    
```

```python
# while loop
count = 0
while count < 5:
    print("Count is", count)
    count += 1

# break and continue
for n in range(1, 10):
    if n == 7:
        break          # stop the loop entirely
    if n % 2 == 0:
        continue        # skip even numbers
    print(n, end=" ")
print()

# else clause on a loop -- runs only if the loop was NOT broken
for n in range(2, 10):
    if n % 7 == 0:
        print(f"Found multiple of 7: {n}")
        break
else:
    print("No multiple of 7 found")
```

## 7. Functions

Python functions are reusable blocks of code used to perform a specific task. 

### Function arguments 
Arguments are values passed to a function when it is called

    - Default -> use a predefined value when no value is passed during the function call.
    - keyword ->  pass values using parameter names, so argument order does not matter.
    - positional -> values are assigned to parameters based on their order in the function call.
    - arbitary -> allow function to accept multiple values. 
          - *args ->  collects extra positional arguments as a tuple.
          - **kwargs -> collects extra keyword arguments as a dictionary. 


### Pass by Reference and Pass by Value
Variables refer to objects. Function behavior depends on whether the object is mutable or immutable.

    - Mutable objects like lists can be modified inside functions.
    -Immutable objects like integers and strings remain unchanged.


```python
def greet(name):
    """Return a greeting for the given name."""
    return f"Hello, {name}!"

print(greet("Intern"))

# Default arguments
def power(base, exponent=2):  
    return base ** exponent

print(power(5))        # uses default exponent=2
print(power(5, 3))     # overrides default 

# Keyword arguments (order doesn't matter when named)
def describe_person(name, age, city):
    print(f"{name}, {age}, from {city}")

describe_person(age=25, city="Ahmedabad", name="Riya")

def nameAge(name, age):
    print("Hi, I am", name)
    print("My age is ", age)

print("Case-1:")
nameAge("Olivia", 27) # positional arguments 

# *args and **kwargs -- variable number of arguments
def add_all(*numbers):
    return sum(numbers)

def print_info(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

print(add_all(1, 2, 3, 4))
print_info(name="ram", role="Intern", team="AI")




# pass by value and pass by reference 

def myFun(x):
    x[0] = 20

b = [10, 11, 12, 13]  # list is mutable therefore the first value in list is changed 
myFun(b)
print(b)

def myFun2(x):
    x = 20    # integer are immutable so value of a will not change 

a = 10
myFun2(a)
print(a)
```

### Lambda Function 

 is an anonymous (nameless) function that is defined in a single line using the lambda keyword. It is used for short, simple operations where defining a full function is unnecessary.

 - lambda arguments: expression

```python
# single argument 
res = lambda x: x * x
print(res(4))

# multiple argument 
add_three = lambda a, b, c: a + b + c
print(add_three(5, 10, 15))  # Output: 30

# inline conditinal 
check_even = lambda x: "Even" if x % 2 == 0 else "Odd"
print(check_even(7))  # Output: Odd
```

### scope 
- local scope -> A variable created inside a function belongs to the local scope of that function, and can only be used inside that function. it is available for any function inside the function.
- Global scope -> A variable created in the main body of the Python code is a global variable and belongs to the global scope.

- global keyword -> if you need to create a global variable, but are stuck in the local scope, you can use the global keyword.
- nonlocal keyword -> The nonlocal keyword is used to work with variables inside nested functions.

```python
# local scope 
def myfunc():
  x = 300
  print(x)
myfunc()

# global scope
x = 300
def myfunc():
  print(x)
myfunc()
print(x)

# global keyword 
def myfunc():
  global x
  x = 300
myfunc()
print(x)

# nonlocal keyword 
def myfunc1():
  x = "Jane"
  def myfunc2():
    nonlocal x
    x = "hello"
  myfunc2()
  return x

print(myfunc1())
```

```python
def fun(lst=[]):
    lst.append(1)
    return lst

print(fun())
print(fun())
print(fun())
print(fun())
print(fun())

# Default arguments are created once, when the function is defined—not every time it is called.
```

```python
def fun(lst=None):
    if lst is None:
        lst = []
    lst.append(1)
    return lst

print(fun())
print(fun())
print(fun())
```

```python
def fun(a, b=[]):
    b.append(a)
    return b

print(fun(1))
print(fun(2))
print(fun(3, []))
print(fun(4))
```

```python
# Scope: local vs global
counter = 0   # global variable

def increment():
    # this creates a NEW local variable, does not touch the global one
    counter = 10
    counter += 1
    print("Inside function:", counter)

increment()
print("Outside function:", counter)   # still 0

def increment_global():
    global counter
    counter += 1     # now modifies the global variable

increment_global()
print("After increment_global:", counter)
```

```python
# Lambda functions -- small anonymous functions
square = lambda x: x ** 2
print(square(6))

# Commonly used with map/filter/sorted
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
doubled = list(map(lambda x: x * 2, nums))
print("Evens:", evens)
print("Doubled:", doubled)

people = [("Alice", 30), ("Bob", 25), ("Charlie", 35)]
sorted_by_age = sorted(people, key=lambda p: p[1])
print(sorted_by_age)
```
