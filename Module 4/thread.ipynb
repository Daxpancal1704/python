```python
"Many Threade with a loop"

def match_schedule(f_team,s_team):
    print(f" Team name : {f_team} vs {s_team}")
    time.sleep(4)

f_teams=["ind","aus"]
s_teams=["wi","eng"]

start=time.perf_counter()

threads=[]

for f_team,s_team in zip(f_teams, s_teams):

    t=threading.Thread(target=match_schedule,args=(f_team, s_team))
    threads.append(t)
    t.start()
for t in threads:
    t.join()

end = time.perf_counter()
print(f"All matches scheduled in {end - start:.2f} seconds")
```

```python
def match_s(match):
    print(f"match is {match}")
    time.sleep(3)
    print(f"end match is {match}")

matches = ["IND vs WI", "AUS vs ENG", "SA vs NZ"]

start = time.perf_counter()
threads = [threading.Thread(target=match_s, args=(m,)) for m in matches]

for t in threads:
    t.start()

for t in threads:
    t.join()

print(f"{len(matches)} scorecards in {time.perf_counter() - start:.2f} seconds")
```

```python
import threading
import time


def cpu_heavy(n):
    total=0
    for i in range(n):
        total += i*i
    return total

N=10_000_000
TASK=4

# sequential executions
start=time.perf_counter()

for _ in range(TASK):
    cpu_heavy(N)

sequential_time=time.perf_counter() - start

print(f"Sequential time:{sequential_time:.2f} s")


# Multithreading execution

start=time.perf_counter()
thread=[threading.Thread(target=cpu_heavy ,args=(N,))for _ in range(TASK)]
for i in thread:
    i.start()
for i in thread:
    i.join()
print(f"4 thread time {time.perf_counter() - start:.2f} s")
```

```python
"Race Conditions -> A race conditions occurs when two or more thread access and modify sheard data at the same time ,and the final result occures on order of timing of thier executions. "


import time
import threading

total_runs=0

def score_r(balls):

    global total_runs

    for _ in range(balls):
        current=total_runs
        time.sleep(0)
        total_runs=current +1

scores=[threading.Thread(target=score_r,args=(1000,))
for _ in range(4)]
for t in scores:
    t.start()
for t in scores:
    t.join()

print("Expected runs: ",5*1000)
print("Actual run : ",total_runs)
```

```python
"Lock -> lock refers to  a threading lock, which is used in multithreding to prevent multiple thread from accessing  sheard data at the same time. "
"A lock let only one thread at a time into a block"


import threading
import time 

total_runs=0
pen=threading.Lock()

def score_run_safe(balls):
    global total_runs
    for _ in range(balls):
        with pen:
            current=total_runs
            time.sleep(0)
            total_runs=current + 1
            

score=[threading.Thread(target=score_run_safe , args=(1000,)) for _ in range(5)]
for t in score: t.start()
for t in score:t.join()

print("Expected runs : ",5*1000)
print("Actual runs : ",total_runs)



```
