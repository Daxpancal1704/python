```python
# %%writefile demo.py
import multiprocessing
import os
import time


"Multiprocessing -> is the use of two or more center processing unit(CPUs) or processor core within a single computer  system to execute  multiple processes simultaneously."

def play_matc(match,seconds):
    print(f"{match} started in process {os.getpid()}")
    time.sleep(seconds)
    print(f"{match} finsihed")

if __name__=="__main__":
    print("Main preocess id : ",os.getpid())
    p1=multiprocessing.Process(target=play_matc,args=("Ind vs aus ",1))
    p2=multiprocessing.Process(target=play_matc, args=("enf vs sa",1))

    start=time.perf_counter()
    p1.start()
    p2.start()
    p1.join()
    p2.join()
    print(f"Both matche done in {time.perf_counter() - start:.2f} s")
```

```python
"Pool executors -> both pool and executors  helps  as run  multiple  tasks using multi processor and thread without manually creating every process or thread. "


# 1.What is pool in python
" pool is provider by the  mutliprocessing module"
" it creates a  group of worker processes. insetead of creating processes manually , you give task to the pool ,and it distributes the work among avilable workers."


import multiprocessing
import time

def square(number):
    print(f"processing {number}")


    time.sleep(2)
    return number * number

if __name__=="__main__":

    numbers=[1,2,3,4,5]
    
    with multiprocessing.Pool(processes=3)as pool:
        result=pool.map(square,numbers)
    print("result :",result)

```

```python
import multiprocessing
import time

def square(number):
    print(f"Processing {number}")

    time.sleep(2)

    return number * number


if __name__ == "__main__":

    numbers = [1, 2, 3, 4, 5]

    with multiprocessing.Pool(processes=3) as pool:

        results = pool.map(square, numbers)

    print("Results:", results)
```
