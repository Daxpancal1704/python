* Abstract base class 
* abstract base clss is used to archieve data abstractions  by defining a common interface for its subclass.

```python
from abc import ABC,abstractmethod

class greet(ABC):
    @abstractmethod
    def say_hello(self):
        pass # this is my abstract methods

class english(greet):
    def say_hello(self):
        return "Helloo"

e=english()
print(e.say_hello())
```
