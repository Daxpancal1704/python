* <b>Creating a Custom iter</b>
* Creating a custom iter defining class that implement the __iter__() and __next__() methods accordng to the python iterator protcol.
* <b>steps</b> :
* Define the class
* initialize attributes: in the __init__() methods of the class 
* implment __iter__() : this methods return the iterator object itself. this is usually as simple returning self.
* implement __next__() : this method should provide th next item in the  sequence each time it's called.

```python
class EvenNumber:
    def __init__(self,limit):
        self.limit=limit
        self.n=2

#return the iter object
    def __iter__(self):
        return self 

# this methods provide the next item in iter
    def __next__(self):
        if self.n > self.limit:
            raise StopIteration

        x=self.n
        self.n+=2
        return x

even=EvenNumber(100)

for num in even:
    print(num,end=" ")
```

```python
"StopIteration Exception"

li=[100,200,00,3333,22,2,343,4,32]
it=iter(li)

while True:
    try:
        print(next(it))
    except StopIteration:
        print("end")
        break
```
