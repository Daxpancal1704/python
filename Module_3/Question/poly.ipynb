* <b> Compile time polymorephism</b>
* This code demonstrates  overloading-like  behaviour using  vairbale length argument 

```python
class calculator:
    def add(self,*num):
        return sum(num)

cal=calculator()
print(cal.add(10))
print(cal.add(10,20))
print(cal.add(10,20,30))
print(cal.add(10,20,30,40))
```

* <b>Runtime polymorephism</b>
* The followin code demonstrate runtime polymorphism using method overriding.
* same method name  but differnt class name 


```python
class Animal:
    def sound(self):
        print("Animal sound")

class dog(Animal):
    def sound(self):
        print("dog sound")

class cat(Animal):
    def sound(self):
        print("cat")

Aniamls=[dog(),cat(),Animal()]

for ani in Aniamls:
    ani.sound()
    
```
