* Coupling : Not Depending too  Much on other 
Coupling is about how tangled two different pieces of code are. alway want to "low coupling."
* Bad(High coupling):A waiter who has to run into the kitchen ,grab the pan, and cook the food themslevesif the stove changes , the waiter cannot do their job.
in pythonwhen the one function directly reaches inside another class and change its private class
* good (low coupling) : a waiter who simply  hands a written  order ticket to the chef . the waiter does nit care how the  chef cooks. 
in python function just pass simple data (like number or text) to each other  


```python
# BAD: High Coupling (print_bill directly depends on a specific dictionary layout)
def print_bill(user_profile):
    # If 'billing' or 'total' key names ever change, this breaks!
    print(f"Total: ${user_profile['billing']['total']}")

# GOOD: Low Coupling (print_bill only asks for the exact number it needs)
def print_bill(total_amount):
    print(f"Total: ${total_amount}")
```
