```python
from collections import Counter

a=[1,2,3,2,3,2,34,5,5,4,4,5,5]

count=Counter(a)
print(count)



countstr=Counter("Hello My name is dax")
print(countstr)


```

```python
#Accessing Counter element
from collections import Counter

ctr=Counter([1,2,1,2,3,2])
ctr1=Counter(["Hello"])
print(ctr[1])
print(ctr1[1])
```

```python
#update

ctr=Counter([1,2,3])
ctr.update([1,2,3])
print(ctr)
```

```python
#element() return an iterator  over element repating each as many times as its count.

ctr=Counter([1,2,34,5,6,7,8,9,8,9])
items=ctr.elements()
print(set(items))
```

```python
#most_common(): return the most common elements their count from the most common of the list 
#most_common(2) -> use for how many number you get most comman mens 2 most coman number you get
ctr=Counter([1,2,12,3,4,3,4])
commonel=ctr.most_common(3)
print(commonel)
```

```python
#Increasing the count of  a single elemnet by 1.

ctr=Counter([1,2,3])

ctr[1] += 10
ctr[2] += 10
ctr[3] += 10
ctr[4] += 10

print(ctr)
```

```python
#subtract() -> subtract eement count  from another iterable  or maping 

ct=Counter([1,2,3,4])
ct.subtract([1,2])
print(ct)
```

```python
#Arithmetic operations

ctr1=Counter([1,2,3,4,4])
ctr2=Counter([1,2,3,4,4])

print(ctr1+ctr2)
print(ctr2-ctr1)
print(ctr2 | ctr1)
```
