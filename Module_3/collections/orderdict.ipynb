```python
from collections import OrderedDict

od=OrderedDict()
od["apple"]=2
od["apples"]=2
print(list(od.items()))
```

```python
# OrderedDict vs dict

d={}

d["asw"]=1
d["accec"]=21
d["adc"]=2

for keys ,values in d.items():
    print(keys,values)
print(d)
print(OrderedDict(reversed(d.items())))
```

```python
#pop last for first item

d=OrderedDict([('a',1),('b',2),('c',3)])
res=d.popitem(last=True)
print(res) #removing last item
print(d)
```

```python
#Moves key to front to end
d=OrderedDict([('a',1),('b',2),('c',3),('d',4)])
d.move_to_end("a")
# d.move_to_end('b')
for k ,v in d.items():
    print(k,v)
```
