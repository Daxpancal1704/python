* A deque stands  for double-Ended Queue. it is a type of data structure that allows to 'add  remove  element  from both  ends ' efficieny.
* allow fast insertion  and delations both end front and rear ends
* work both queue (LIFO) and stack(FIFO).
* use full from in scheduling ,sliding window problem and real time data processing

```python
from collections import deque

d=deque([12,23,34])
print(d)
```

```python
#Accessing item in deque:
try:
    d=deque([1,2,3,4,"Alice","bob"])
    print(d[1])
    print(d[1])
    print(d[5])
    print(d[-6])
except IndexError:
    print("Error")    
```

```python
#Opearation on deque
#"1.append()"

d=deque([1,2,3,4,"A",'V'])

d.append("Dax")
print(list(d))

#"2.appendleft"

d.appendleft("Panchal")
print("left side append element :",list(d))

#"extend()-> add a multiple element right of the deque."
d.extend(["Alice",10,"Jaen Foster"])
print("extend() use -> Add element of the rigth of the deque",list(d))

#extendleft() ->  add multiple element left og the deque

d.extendleft([00,-1,12,234])
print("extendleft()-> use for add multiple elemnet left side",list(d))

# remove() -> remove the first occurrence of a specified value.

d.remove("Panchal") 
print("Remove the value ",list(d))

#pop() reomve the value and return value

print("before :",list(d))
d.pop()
print("after :",list(d))

#popleft() -> reomve value from left side 

d.popleft()
print("remove value from left side",list(d))


#len()-> check the len for deque

print("check len :",len(d))

#cout()-> count words
print("Count : ",d.count(10))

#rotate()-> rotate element 
print("Before rotate element :",list(d))
d.rotate(10)
print("after rotate element :",list(d))

#reverse()-> element
d.reverse()
print("reversed element : ",list(d))
#clear() -> use for clear the deque
d.clear()
print(d)
```
