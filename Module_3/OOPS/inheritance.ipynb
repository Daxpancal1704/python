# It is allows to new class(child or derived class) to inherit attributes or methods from exisiting class.

```python
class Animal:
    def __init__(self,name):
        self.name=name
    def info(self):
        print("Animal name : ",self.name)

class Dog(Animal):
    def voice(self):
        print(self.name)

d1=Dog("BHav")
print(d1.name)
d1.info()
d1.voice()
```

```python
"# super() functions is used to call methods from parent(superclass) inside a chid (subclass)."
#single inheritance
class parent:
    def show1(self):
        print("Hello this is paret class and show methods")
class child(parent):
    def show(self):
        super().show1()#calling parent methods
        print("This is child class methods")
c1=child()
c1.show()

```

```python
"Multiple inheritance"

class class1:
    print("Hello")
    def __init__(self,name,*args,**kwargs):
        self.name=name
        super().__init__(*args,**kwargs)
        print("This is clss1 ",self.name)
class class2:
    def __init__(self,*args,**kwargs):
       super().__init__(*args,**kwargs)
       print("This is class2 ",self.name)

class class3(class1,class2):
   pass

c3=class3("Alice")
print(c3)
```

```python
"Multiple inheritance"



class demo1:
    def d1(slef):
        print("Hello")
class demo2(demo1):
    def d2(slef):
        print("good")
class demo3(demo2):
    def d3(slef):
        print('morning')

s1=demo3()
s1.d1() 
s1.d2()
s1.d3()
```

```python
"Hierarchical Inheritance : - Multiple child classes inherit from th same parent class."

class Animal:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def eat(self):
        print(self.name,"is eating")
class Dog(Animal):
    def __init__(self, name,age,breed):
        super().__init__(name,age)
        self.breed=breed

    def bark(self):
         print(self.name, "is barking")
        

class cat(Animal):
    def __init__(self, name, age, color):
            super().__init__(name, age)
            self.color = color
    
    def meow(self):
        print(self.name, "is meowing")         

d1= Dog("Bruno", 5, "Labrador")
c1=cat("Simba",3,"Orange")

print(d1.name)
print(c1.name)
print(c1.age)
print(c1.color)
```

```python
"Method resolution order"


class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show_details(self):
        print("Name: ",self.name)        
        print("Age: ",self.age)



class student(Person):
    def __init__(self, name, age,roll_no):
        Person.__init__(self,name, age)
        self.roll_no=roll_no

    def show_details(self):
        print("Show Roll No: ",self.roll_no)
        super().show_details()

class Employee(Person):
    def __init__(self, name, age, emp_id):
        Person.__init__(self,name, age,emp_id)
        self.emp_id=emp_id

    def show_details(self):
        print("Empd id :",self.emp_id)
        super().show_details()


class TeachingAssistant(student,Employee):
    def __init__(self, name, age, roll_no,emp_id):
        student.__init__(self, name, age, roll_no)
        self.emp_id=emp_id


    def show_details(self):
         super().show_details()

t1=TeachingAssistant("Alice",21,101,11)
t1.show_details()


        
```
