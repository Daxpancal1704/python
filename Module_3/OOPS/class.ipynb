```python
"__repr__ methods use to define more technical represitions,menat for developers"
"__repr__ methods calling the repr() when use bothe __str__ and __repr__ "


class demo:
    def __init__(self,name,age):
        self.name=name
        self.age=age


    def __str__(self):
        return f"{self.name!r} : {self.age}"

    def __repr__(self):
        return f" prson(name={self.name!r},age={self.age})"

d1=demo("ALice",23)
print(d1)
print(repr(d1))
```

```python
"Class Methods @classmethod used to create a methods that work with the class insted of object instance class methoids recived th class itself as the first argument "

"cls referse to a class itself "
"args it is optional to defind"

class student:
    total_student=0

    def __init__(self,name):
        self.name=name
        student.total_student+=1

    @classmethod
    def get_total_stud(cls):
        return f"Total students count is {cls.total_student}"
s1=student("alice",)
s2=student("alice")
s1.get_total_stud()    
s2.get_total_stud()    
```

```python
"@staticmethods"
"A static method in python is a method defines inside  a class that does not deppend on any instance or class data "
"Does not recive any self keyword"


class demp:
    def __init__(self,age):
        self.age=age

    @staticmethod
    def fun(age):
            return age >= 18
        
print(demp.fun(20))
d=demp(40)
print(d.fun(d.age))

```
