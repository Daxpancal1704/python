```python
"Polymorephism means 'many forms' it is allows to diffrent types of class og object accssed througth the same unified interface . this means singe function , methods, opertor can behave differntly depending on object acting"

"there are two types of polymorephism"
"1.Compile-time : it is allows multiple methods as same name but different parmeter "
"2. Runtime polymorphism : it means same methid an different class name "
```

```python
"Compile-time polymorphism"


class Cal:
    def mul(self,a,b,*args):
        res=a+b
        for num in args:
            res+=num
        return res
c1=Cal()
print(c1.mul(1,2,3))
print(c1.mul(10,2,3))
print(c1.mul(10,2,3,2,3,4))
```

```python
"Run-time polymorephism "

class Person:
    def pay(self,name,amount):
        self.name=name
        self.amount=amount
        print(name,amount)
        print()

class Adult(Person):
    def pay(self,age):
        self.age=age
        print(age)
p1=Person()
a1=Adult()
p1.pay("Alice",10000)   
a1.pay(23)   

```
