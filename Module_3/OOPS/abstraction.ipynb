```python
"Can not create a direactly greet object beaause greet is abstract methods   "

from abc import ABC , abstractmethod

class greet(ABC):
    @abstractmethod
    def say_hello(self):
        pass

    def say_by(self):
        return "Byy"

class english(greet):
    def say_hello(self):
        return "Hello!"

e=english()
e.say_hello()
e.say_by()
```

```python


from abc import ABC, abstractmethod


# Abstract class
class Payment(ABC):

    @abstractmethod
    def payment(self, amount):
        pass

    def receipt(self):
        return "Payment completed successfully!"


# Child class 1
class CreditCard(Payment):

    def payment(self, amount):
        return f"Paid Rs.{amount} using Credit Card"


# Child class 2
class UPI(Payment):

    def payment(self, amount):
        return f"Paid Rs.{amount} using UPI"


# Child class 3
class Cash(Payment):

    def payment(self, amount):
        return f"Paid Rs.{amount} using Cash"


# Creating objects
p1 = CreditCard()
p2 = UPI()
p3 = Cash()


# Calling methods
print(p1.payment(1000))
print(p1.receipt())

print(p2.payment(500))
print(p2.receipt())

print(p3.payment(200))
print(p3.receipt())
```
