# Access Modifiers
* 1. Public : public members are vairabel or methods that can be accesssed from anywhere inside the class ,outside class or from other modules.
By default all memeber are public.without any underscore prefix,"self.name".
* 2. Protected : protected members are vairble or methods that are intende to be access only with in the class ans subclass.define by "_" single underscore prefix, ex. "self._age"
* 3. Private :Privte members are vairble or methods that cannot be accessed directly from outside the class . defined by double uderscore "__" prefix.
"self.__marks"   

```python
class student:
    def __init__(self,name,age):
        self.name=name #public 
        self._age=age #protected
        self.__marks=89 #private

        # Public methods

    def set_age(self,age):
            try:
                if age < 0:
                    raise ValueError("Age cannot be negative")
                self._age=age

            except ValueError as e:
                print("Error:",e)
        
        #Public Methods

    def get_age(self):
            return self._age

    def get_marks(self):
            return self.__marks

s=student("Alice",23)

print("Name :",s.name)
print("Age : ",s._age)

#private member
print("Marks ",s.get_marks())


s.set_age(77)
print("Age : ",s.get_age())

# Invalid value
s.set_age(-10)
print("Age:", s.get_age())

```
