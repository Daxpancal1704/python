### Object-Oriented Programming in Python 

<details>
<summary>What is OOP's?</summary>

OOP (Object-Oriented Programming) is a programming paradigm that organizes software using classes and objects, combining data and methods into a single unit.
</details>

<details>
<summary> Why we use OOP's</summary>

1. <b>Code Reusability</b> – Allows existing code to be reused through classes and objects, reducing the need to write the same code repeatedly.
2. <b>Better Code Organization</b> – Groups related data and functions together within a class, making the code more structured.
3. <b>Easy Maintenance</b> – Makes the program easier to understand, modify, debug, and maintain.
4. <b>Less Code Duplication</b> – Promotes the reuse of existing code, avoiding unnecessary repetition.
5. <b>Data Security</b> – Encapsulation helps protect data by controlling how it can be accessed and modified.
6. <b>Managing Complex Programs</b> – Divides a large and complex program into smaller, independent, and manageable classes.
7. <b>Scalability</b> – Makes it easier to extend the application and add new features without significantly changing existing code.
</details>

<b>ClassMethod</b> - It a decorator which allows us to convert any method into a class method

````python
class Cricketer():
    nationality="Australian" 
    def __init__(self,name):
        self.name = name
    
    @classmethod
    def country(cls):       #cls is a variable like self we can use any different variable name if we want
        return cls.nationality      
    
c1=Cricketer("Steve Smith")
print(c1.name,Cricketer.country())
````

````python
class GFG:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f'Name is {self.name}'

    def __repr__(self):
        return f'GFG(name is{self.name})'


obj = GFG('GeeksForGeeks')

print(type(obj.__str__()))
print(obj.__repr__())
````

<b>StaticMethod</b> - It is a class method but it doesn't require any access instance or class parameter. We don't need to create the object for staticmethods.

Encapsulation is the bundling of attributes and methods within a class, restricting access to some components to control interactions. 
</details>

<details>
<summary>Why do we need Encapsulation?</summary>

1. Improves maintainability, readability, and usability
    - Encapsulation allows the methods of a class to directly access its data members, so we do not need to explicitly pass the same data as arguments every time.
2. Maintains data integrity
    - Encapsulation allows us to control how data is accessed or modified. We can apply validation before assigning values, which helps prevent invalid or unwanted data.
3. It helps in achieving abstraction.
    - Encapsulation allows a class to hide its internal implementation details and expose only the required functionality to other classes.
4. Improves security
    - Sensitive or important data can be restricted from direct access.
</details>

##### Access Modifiers

1. <b>Public</b> - Public members are variables or methods that can be accessed from anywhere inside the class, outside the class or from other modules. By default, all members are public. defined without any underscore prefix, "self.name".
2. <b>Protected</b> - Protected members are variables or methods that are intended to be accessed only within the class and its subclasses. defiend by a "_" single underscore prefix , "self._age". 
3. <b>Private</b> - Private members are variables or methods that cannot be accessed directly from outside the class. defined by double underscore "__" prefix, "self.__marks"
 

````python
class Student:
    def __init__(self, name, age):
        self.name = name          # Public
        self._age = age           # Protected
        self.__marks =  82         # Private

    # Public method
    def set_age(self, age):
        try:
            if age < 0:
                raise ValueError("Age cannot be negative")

            self._age = age

        except ValueError as e:
            print("Error:", e)

    # Public method
    def get_age(self):
        return self._age

    # Public method to access private data
    def get_marks(self):
        return self.__marks


s = Student("Sachit", 21)

# Public member
print("Name :", s.name)

# Protected member
print("Age :", s._age)

# Private member
print("Marks from getter method: ",s.get_marks())

# Modifying age through controlled access
s.set_age(22)
print("Age:", s.get_age())

# Invalid value
s.set_age(-10)
print("Age:", s.get_age())
````

<details>
<summary> 4. <b>Inheritance</b> </summary >
Inheritance is an OOP mechanism in which a class(child or derived class), acquires the attributes and methods of an existing class(parent or base class). The child class can reuse, extend, or override the functionality inherited from the parent class.
</details>

<details>
<summary>Advantages of Inheritance?</summary>

1. Promotes code reusability by allowing child classes to reuse the attributes and methods of a parent class
2. Simplifies maintenance by keeping common functionality in the parent class.
3. Enables method overriding by allowing a child class to provide its own implementation of a method inherited from the parent class.

<b>super()</b> : is a built-in function used in inheritance to access the methods and constructor of a parent class from a child class, while following Python's Method Resolution Order (MRO).

##### Types of Inheritance

1. Single Inheritance -  One parent and one child class

````python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no


s = Student("Sachit", 101)

print(s.name)
print(s.roll_no)
````

2. Multi-Level Inheritance -  In this a class is child of a parent class and this child class is then a parent of another child class

````python
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    def bark(self):
        print("Barking")


class Puppy(Dog):
    def play(self):
        print("Playing")


p = Puppy()
p.eat()
p.bark()
p.play()
````

3. Multiple Inheritance - When for a single child class there is two or more than two parent class

````python
class Father:
    def __init__(self, father_name):
        self.father_name = father_name


class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name


class Child(Father, Mother):
    def __init__(self, father_name, mother_name, child_name):
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
        self.child_name = child_name


c = Child("John", "Mary", "David")

print(c.father_name)
print(c.mother_name)
print(c.child_name)
````

4. Hierarchical Inheritance - Multiple child classes inherit from the same parent class.

````python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def eat(self):
        print(self.name, "is eating")


class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)
        self.breed = breed

    def bark(self):
        print(self.name, "is barking")


class Cat(Animal):
    def __init__(self, name, age, color):
        super().__init__(name, age)
        self.color = color

    def meow(self):
        print(self.name, "is meowing")


d = Dog("Bruno", 5, "Labrador")
c = Cat("Simba", 3, "Orange")

print(d.name)
print(d.age)
print(d.breed)

d.eat()
d.bark()

print(c.name)
print(c.age)
print(c.color)

c.eat()
c.meow()
````

5. Hybrid Inheritance - It is a mix of 2 types of inheritance

````python
class Student:
    def show(self):
        print("Student")


class Employee:
    def show(self):
        print("Employee")


class TeachingAssistant(Employee,Student):
    pass

ta = TeachingAssistant()
ta.show()
````

````python
class Person:
    def __init__(self, name):
        self.name = name

    def show_details(self):
        print("Name:", self.name)


class Student(Person):
    def __init__(self, name, roll_no):
        Person.__init__(self, name)
        self.roll_no = roll_no

    def show_details(self):
        print("Student Details")
        print("Roll No:", self.roll_no)
        super().show_details()


class Employee(Person):
    def __init__(self, name, employee_id):
        Person.__init__(self, name)
        self.employee_id = employee_id

    def show_details(self):
        print("Employee Details")
        print("Employee ID:", self.employee_id)
        super().show_details()


class TeachingAssistant(Student, Employee):
    def __init__(self, name, roll_no, employee_id):
        Student.__init__(self, name, roll_no)
        self.employee_id = employee_id

    def show_details(self):
        print("Teaching Assistant Details")
        super().show_details()


ta = TeachingAssistant("Sachit", 101, 5001)

ta.show_details()
````

<details>
<summary> 5. <b>Abstraction</b> </summary >

Abstraction is the process of hiding implementation details and exposing only the essential functionality to the user. It is used to hide the implementation details from the user and expose only necessary parts. 
- It is acheived using <b> @abstractmethod </b> decorator
</details>

````python
from abc import ABC , abstractmethod
# Abstract Base Class (ABC)
class Car(ABC):                     # This is creating Abstract class inside which abstract method is defined                   
    def __init__(self,model,price,brand):
        self.brand=brand
        self.price=price
        self.model=model

    @abstractmethod                 # untill Abc is not inherited you can't create a abstract class even if you have written @abstractmethod
    def print_details(self):        # Abstract propeties, - must be implemented by the subclass
        return f"{self.model}"  

    def get_brand(self):            # Concrete methods - can be directly accessed and used by sib classes
        print(self.print_details())
        return self.brand
    
class Hatchback(Car):  
    def print_details(self):
        print("hello")
        return f"{self.model} {self.price}"
    
h1=Hatchback("Q7",9000000,"Audi")
print(h1.print_details(),h1.get_brand())
````

<details>
<summary> 6. <b>Polymorphism</b> </summary >

Polymorphism allows the same method, function or operator to behave differently depending on the object or data it works with. This flexibility helps create more reusable, maintainable and scalable code.

Types of Polymorphism in python

1. Compile-time Polymorphism 
2. Run-time Polymorphism

````python
## Method overloading 
class Calculator:
    def multiply(self, *args):
        result = 1
        for num in args:
            result *= num
        return result

# Create object
calc = Calculator()
        
# Using multiple arguments
print(calc.multiply(2, 3))       
print(calc.multiply(2, 3, 4))
````

````python
#Runtime Polymorphism
class Add():
    def __init__(self,a):
        self.a=a
    def __add__(self,other):
        return self.a+other.a
    def __sub__(self,other):
        return self.a-other.a
    def __mul__(self,other):
        return self.a * other.a
    
a1=Add(2)
a2=Add(7)
print(a1*a2)   
print(a2-a1) 
````

1. <b>Cohesion </b> - The degree to which the responsibilities within a class or module are closely related.

2. <b>Coupling </b> - The degree of dependency between different classes or modules.

````python
##Example
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount

    def get_balance(self):
        return self.balance


class NotificationService:
    def send_notification(self, message):
        print("Notification:", message)


class Bank:
    def __init__(self, notification_service):
        self.notification_service = notification_service

    def process_deposit(self, account, amount):
        account.deposit(amount)

        self.notification_service.send_notification(
            "Deposit successful"
        )


notification = NotificationService()

bank = Bank(notification)

account = BankAccount("ACC101", 5000)

bank.process_deposit(account, 2000)

print("Balance:", account.get_balance())
````

#### Memory Management in Python

<details>
<summary>What is Memory Management?</summary>

Python automatically allocates memory when an object is created and frees it when the object is no longer needed, using <b>reference counting</b> (memory is freed once no variable refers to the object) and a <b>garbage collector</b> (cleans up objects stuck in reference cycles). The `__del__` method can be defined to run cleanup code just before an object is destroyed. 
</details>

````python
class Connection:
    def __init__(self, name):
        self.name = name
        print(f"{self.name} connection opened")

    def __del__(self):                        # called automatically when object is destroyed
        print(f"{self.name} connection closed")

conn = Connection("Database")
del conn                                       # removes the reference -> triggers __del__
````

#### Lambda Functions & Functional Programming

<details>
<summary>What is a Lambda Function?</summary>

A lambda function is a small, anonymous (unnamed) function defined using the `lambda` keyword instead of `def`. It can take any number of arguments but has only one expression, which is automatically returned. Useful for short, throwaway functions, especially with `map`, `filter`, and `reduce`.
</details>

````python
square = lambda x: x ** 2
print(square(5))
````

1. map()

Applies a function to every item of an iterable and returns a map object (convert to `list` to view).

````python
nums = [1, 2, 3, 4, 5]
squares =  (map(lambda x: x ** 2, nums))
print(list(squares))
````

2. filter()

Keeps only the items of an iterable for which the given function returns True.

````python
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
````

3. reduce()

Repeatedly applies a function to pairs of items in an iterable, reducing it to a single value. Needs to be imported from functools.

````python
from functools import reduce

nums = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, nums)
print(product)      # 1*2*3*4*5
````

##### Always write patterns as raw strings: `r"..."`
In a normal string Python reads backslashes itself (`"\n"` = new line, `"\b"` = backspace). A **raw string** `r"..."` hands the backslashes untouched to the `re` module, which is what we want for `\d`, `\w`, `\b` ... From here on every pattern is written as `r"..."`.

````python
print(len("\b"), len(r"\b"))                      # 1 character (backspace) vs 2 characters (\ and b)

text = "cat and concat"
print(re.findall("\bcat\b", text))                # normal string: \b became a backspace -> no match
print(re.findall(r"\bcat\b", text))               # raw string: \b is a word boundary -> only the word cat
````

<summary>Character classes</summary>

| Pattern | Meaning | Output |
|---|---|---|
| `\d` | Any digit (0-9) | `1, 8, 3, 2, 6, 4, 2, 0, 8` |
| `\D` | Any non-digit | All characters except digits |
| `\w` | Any word character (letters, digits, underscore) | All letters and digits |
| `\W` | Any non-word character | Spaces and `+` |
| `\s` | Any whitespace (space, tab, newline) | All spaces |
| `\S` | Any non-whitespace | All non-space characters |
| `.` | Any character except newline | Every character in the string |
| `[abc]` | Any one of `a`, `b`, or `c` | `a, a, a, a, a, b, a` |
| `[^abc]` | Any character except `a`, `b`, or `c` | All characters except `a`, `b`, `c` |
| `[a-z]` | Any lowercase letter | All lowercase letters |

| `*` | Matches 0 or more occurrences of the preceding pattern | With `\d*`: `183, 264, 208` and empty matches |
| `?` | Matches 0 or 1 occurrence of the preceding pattern | With `\d?`: each digit is matched optionally |
| `{n}` | Matches exactly `n` occurrences of the preceding pattern | With `\d{3}`: `183, 264, 208` |
| `{n,m}` | Matches between `n` and `m` occurrences | With `\d{2,3}`: `183, 264, 208` |
| `\|` | OR operator, matches either pattern on its left or right | With `Virat\|Rohit`: `Virat, Rohit` |

11. () -> Capture group ,it creates the search pattern into different groups

What it does is,it divides the search pattern into different parts and we can use a specific pattern from the whole pattern using capture group and to use a specific group it has indices [0] is whole pattern 
[1] is first part of the pattern [2] is second part and so on

````python
import re

text = "Date: 2025-06-16"
pattern = r"(\d+)-(\d+)-(\d+)"
match = re.search(pattern, text)

if match:
    print("Full match:", match.group(0))  
    print("Year:", match.group(1))       
    print("Month:", match.group(2))      
    print("Day:", match.group(3)) 
````

````python
# groups() returns every group; (?P<name>...) gives a group a name
m = re.search(r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})", "Date: 2025-06-16")
print(m.groups())
print(m.group("year"), m.group("month"))
print(m.groupdict())
````

##### `search()` vs `match()` vs `fullmatch()`

| Function | Where the pattern must match |
| --- | --- |
| `re.search()` | **anywhere** in the string (first match) |
| `re.match()` | at the **start** of the string |
| `re.fullmatch()` | the **whole** string, start to end |

All three return a match object, or `None` when there is no match.

````python
text = "2026-Season"

print(re.search(r"Season", text))            # found anywhere
print(re.match(r"Season", text))             # None - the string does not START with Season
print(re.match(r"\d{4}", text))              # starts with 4 digits
print(re.fullmatch(r"\d{4}-\w+", text))      # the whole string fits the pattern
print(re.fullmatch(r"\d{4}", text))          # None - extra characters are left over
````

##### `finditer()` and `compile()`
`findall()` gives a plain list. `finditer()` gives a match object for every match, so you also get the groups and positions. `compile()` prepares a pattern once so that you can reuse it.

````python
text = "Norris 25 points, Hamilton 18 points, Russell 15 points"

for m in re.finditer(r"(\w+) (\d+) points", text):
    print(m.group(1), m.group(2), m.span())

pattern = re.compile(r"\d+")                     # compile once, use many times
print(pattern.findall("Lap 12 of 57"))
print(pattern.search("Lap 12 of 57").group())
````

##### Greedy vs lazy
`*`, `+`, `?` and `{}` are **greedy** - they match as much as they can. Adding a `?` after them makes them **lazy** - they match as little as possible.

````python
html = "<b>Hamilton</b> and <b>Norris</b>"

print(re.findall(r"<b>.*</b>", html))      # greedy: one big match
print(re.findall(r"<b>.*?</b>", html))     # lazy: each tag on its own
````

##### Putting regex to work - practical patterns

````python
report = """Kimi Antonelli: 219 points, Lewis Hamilton: 169 points, George Russell: 160 points.
Contact press@f1news.com or media.desk@grandprix.org, phone 98765-43210.
Next race: 06-09-2026 at Zandvoort."""

print(re.findall(r"\d+", report))                                   # every number (even the 1 inside f1news!)
print(re.findall(r"[\w.]+@[\w.]+\.\w+", report))                      # e-mail addresses
print(re.findall(r"\d{5}-\d{5}", report))                            # phone number
print(re.findall(r"\d{2}-\d{2}-\d{4}", report))                      # date DD-MM-YYYY
print(re.sub(r"\d{5}-\d{5}", "*****-*****", report).splitlines()[1])  # hide the phone number
````

````python
# groups + a dictionary comprehension: build {driver: points}
pairs = re.findall(r"([A-Z][a-z]+ [A-Z][a-z]+): (\d+) points", report)
print(pairs)

standings = {name: int(points) for name, points in pairs}
print(standings)
````

````python
# a reusable validator built with fullmatch()
def is_valid_email(text):
    return re.fullmatch(r"[\w.]+@[\w]+\.[a-z]{2,}", text) is not None

for e in ["asha.k@example.com", "ravi@mail", "no spaces@x.com", "meera_99@site.org"]:
    print(e, "->", is_valid_email(e))
````

##### Convert the following procedural code into OOP

Create a class `SpaceMission` that stores the mission data as attributes and turns each function into a method (`consume_fuel`, `add_crew_member`, `display_mission_report`). Calling it with the same values must print the same report as the code below.

````python
def manage_space_mission(
    mission_name,
    mission_id,
    destination,
    crew_members,
    fuel_capacity,
    fuel_consumed,
    mission_days
):
    """
    Manage a space mission and calculate its mission status.

    Args:
        mission_name (str): Name of the space mission.
        mission_id (int): Unique identifier of the mission.
        destination (str): Destination of the spacecraft.
        crew_members (list): Names of astronauts assigned to the mission.
        fuel_capacity (float): Total fuel capacity of the spacecraft in liters.
        fuel_consumed (float): Amount of fuel already consumed in liters.
        mission_days (int): Planned duration of the mission in days.

    Returns:
        dict: A dictionary containing mission details, remaining fuel,
              fuel percentage, crew status, and mission status.
    """

    remaining_fuel = fuel_capacity - fuel_consumed
    fuel_percentage = (remaining_fuel / fuel_capacity) * 100

    if fuel_percentage <= 10:
        fuel_status = "Critical"
    elif fuel_percentage <= 25:
        fuel_status = "Low"
    elif fuel_percentage <= 50:
        fuel_status = "Moderate"
    else:
        fuel_status = "Healthy"

    if len(crew_members) == 0:
        crew_status = "No Crew"
    elif len(crew_members) <= 4:
        crew_status = "Small Crew"
    else:
        crew_status = "Large Crew"

    if fuel_percentage <= 10:
        mission_status = "Mission At Risk"
    elif mission_days > 365 and len(crew_members) > 6:
        mission_status = "Long Duration Mission"
    else:
        mission_status = "Mission Operational"

    mission_record = {
        "mission_name": mission_name,
        "mission_id": mission_id,
        "destination": destination,
        "crew_members": crew_members,
        "fuel_capacity": fuel_capacity,
        "fuel_consumed": fuel_consumed,
        "remaining_fuel": round(remaining_fuel, 2),
        "fuel_percentage": round(fuel_percentage, 2),
        "fuel_status": fuel_status,
        "crew_status": crew_status,
        "mission_days": mission_days,
        "mission_status": mission_status
    }

    return mission_record


def consume_fuel(mission_record, fuel_used):
    """
    Consume fuel during the space mission.

    Args:
        mission_record (dict): Current mission information.
        fuel_used (float): Amount of fuel consumed in liters.

    Returns:
        dict: Updated mission record.
    """

    mission_record["fuel_consumed"] += fuel_used

    if mission_record["fuel_consumed"] > mission_record["fuel_capacity"]:
        mission_record["fuel_consumed"] = mission_record["fuel_capacity"]

    return manage_space_mission(
        mission_record["mission_name"],
        mission_record["mission_id"],
        mission_record["destination"],
        mission_record["crew_members"],
        mission_record["fuel_capacity"],
        mission_record["fuel_consumed"],
        mission_record["mission_days"]
    )


def add_crew_member(mission_record, astronaut_name):
    """
    Add an astronaut to the space mission.

    Args:
        mission_record (dict): Current mission information.
        astronaut_name (str): Name of the astronaut to be added.

    Returns:
        dict: Updated mission record.
    """

    if astronaut_name not in mission_record["crew_members"]:
        mission_record["crew_members"].append(astronaut_name)

    return manage_space_mission(
        mission_record["mission_name"],
        mission_record["mission_id"],
        mission_record["destination"],
        mission_record["crew_members"],
        mission_record["fuel_capacity"],
        mission_record["fuel_consumed"],
        mission_record["mission_days"]
    )


def display_mission_report(mission_record):
    """
    Display the complete space mission report.

    Args:
        mission_record (dict): Mission information.

    Returns:
        None: This function only displays the mission report.
    """

    print("\n========== SPACE MISSION REPORT ==========")

    print("Mission Name:", mission_record["mission_name"])
    print("Mission ID:", mission_record["mission_id"])
    print("Destination:", mission_record["destination"])

    print("\nCrew Members:")
    for astronaut in mission_record["crew_members"]:
        print("-", astronaut)

    print("\nFuel Capacity:", mission_record["fuel_capacity"], "L")
    print("Fuel Consumed:", mission_record["fuel_consumed"], "L")
    print("Remaining Fuel:", mission_record["remaining_fuel"], "L")
    print("Fuel Percentage:", mission_record["fuel_percentage"], "%")
    print("Fuel Status:", mission_record["fuel_status"])

    print("\nMission Duration:", mission_record["mission_days"], "days")
    print("Crew Status:", mission_record["crew_status"])
    print("Mission Status:", mission_record["mission_status"])


mission = manage_space_mission(
    "Artemis-X",
    501,
    "Moon",
    ["Alice", "Bob", "Charlie"],
    10000,
    2500,
    180
)

mission = consume_fuel(mission, 1200)

mission = add_crew_member(mission, "David")

display_mission_report(mission)
````

* The balance must **not** be reachable as `acc.__balance` from outside the class.

Make this test code produce the output below:
```python
acc = BankAccount("Asha", 1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.get_balance())

for action, amount in [(acc.withdraw, 5000), (acc.deposit, -50)]:
    try:
        action(amount)
    except ValueError as e:
        print("Error:", e)

try:
    print(acc.__balance)
```
