```python
"lowercase first the uppercase"
"""Given Input: str1 = "PyNaTive"

Expected Output: yaivePNT"""
str1 = "PyNaTive"
l_case=[]
u_case=[]


for char in str1:
    if char.islower():
        l_case.append(char)
    else:
        u_case.append(char)

res="".join(l_case + u_case)
print(res)        

```

```python
"Count all letter,digits,and speical symbols for given string "
"""Given Input: str1 = "P@#yn26at^&i5ve"

Expected Output:

Total counts of chars, digits, and symbols: Chars = 8 Digits = 3 Symbol=4 """


str1="P@#yn26at^&i5ve"
is_digits=0
is_symbol=0
is_char=0

for char in str1:
    if char.isalpha():
        is_char+=1
    elif char.isdigit():
        is_digits+=1
    else:
        is_symbol+=1

print(f"Symbols : {is_symbol} Digits : {is_digits} Characters : {is_char}")        
```

```python
"Calculate the sum and average of the digits present in string"

name="PYnative29@#8496"

count=0
total_digits=0


for char in name:
    if char.isdigit():
        total_digits+=int(char)
        count+=1
average=total_digits/count
print(average)        

```

```python
"Remove the empty string in list"

"""Given Input: str_list = ["Emma", "Jon", "", "Kelly", None, "Eric", ""]

Expected Output: ['Emma', 'Jon', 'Kelly', 'Eric']"""


str_list=["Emma", "Jon", "", "Kelly", None, "Eric", ""]

str_list_new=list(filter(None,str_list))
print(str_list_new)
```

```python
"""
Remove special symbols/punctuation from a string

Given Input: str1 = "/*Jon is @developer & musician!!"

Expected Output: "Jon is developer musician"""
str1 = "/*Jon is @developer & musician!!"


new_string=''
for chars in str1:
    if chars.isalnum() or chars.isspace():
        new_string+=chars
print(new_string)        
```

```python
"""Exercise 25. Find words with both letters and numbers
Given Input: str1 = "Emma25 is Data scientist50 and AI Expert"

Expected Output:

Emma25
scientist50"""


str1 = "Emma25 is Data scientist50 and AI Expert"
res=[]
str11=str1.split()


for char in str11:
    if any(char.isalpha() for char in char)   and any(char.isdigit()  for char in char):
        res.append(char)
for i in res:
    print(i)        
```
