```python
# "Dictionary Methods"
student={
    "name":"Alice",
         "age":23,
         "number":12333
         }
"keys()"
print(student)
print(student.keys())
print(student.values())
print(student.items())

for keys ,values in student.items():
    print(keys,":", values)

student.update(name="Dax")
print(student)    

student.pop(keys)
print(student)
```
