5. Time Converter

Convert a total number of seconds into hours, minutes, and seconds.
Example: 3672 → 1 hour, 1 minute, 12 seconds
Hint: Use // for whole-number division and % for the remainder

```python
second=int(input("second :"))

hours=second//3600
rmaing=second % 3600

minutes=rmaing // 60
second=rmaing % 60


print(hours, "hour,", minutes, "minute,", second, "seconds")
```
