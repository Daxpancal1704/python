Problem 5: Print Identity Matrix Pattern

```python
n=int(input("enter the numbe: "))

for i in range(n):
    for j in range(n):
        if i == j:
            print("1" ,end=" ")
        else:
            print("0",end=" ") 
    print()
```

```python
# Function to print x in increasing order
def printIncreasingPower(x):
    start = 1
    jump = 1

    # Loop to jump in powers of 2
    while jump <= x:
        print(jump, end=" ")
        start += 1
        jump = start**2

printIncreasingPower(100)


# def incres(x):
#     str=1
#     jum=1

#     while jum <=x:
#         print(jum,end="")
#         str=str+1
#         jum=str**2
# incres(10)        
```
