```python
"""python inheritance"""
"Single   inheritance "

class super_class:
    print("good morning")
    name="Alice"

    def __init__(self,name,age):
        self.name=name
        self.age=age

class child_class(super_class):
    print("Hello",super_class.name)
name=input("Name : ")
age=int(input("age : "))


c1=child_class(name,age)
print(f"Hello my name is {c1.name} ,and my age is {c1.age}")

```

```python
"Method Overriding"

class class1 :
    #  name=" "

    def eat(self):
        name="Alice"
        print("Hello this is Superclass eat() ")
        print("Hello",name)

class class2(class1) :
    def eat(self):
        print("Hello thi is a child class eat()")
        super().eat()

cl1=class2()

cl1.eat()
```

```python
"Multiple inheritance Syntax"


class s_class1:
    def  method1(self):
        print("This is First superclass")
class s_class2:
    def method2(self):
        print("This is Second superclass")
class s_class3(s_class1,s_class2):
    def method3(self): 
        print("This is third superclaas")

cla1=s_class3()

cla1.method1()
cla1.method2()
cla1.method3()
```
