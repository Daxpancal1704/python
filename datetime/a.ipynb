```python
#Add hours

from datetime import timedelta
dt=datetime(2026,11,6,14,50)
new_hours=dt+timedelta(hours=2)
print(new_hours)
```

```python
# Timezone class
from datetime import datetime,timezone,timedelta

ist=timezone(timedelta(hours=5,minutes=10),"IST")
dt=datetime(2026,11,6,10,10,tzinfo=ist)
print(dt)
```

# Lambda


```python
name=["Alice","Bob","Joy","Loy"]

ans=list(map(lambda x : len(x),name))
print(ans)

names=["madam","sir","police","teacher","mam"]
res=list(filter(lambda x : x == x[::-1], names))
print(res)
```

</details>

```python
from functools import reduce
nums=[1,2,12,3,4,4]

maximum=reduce(lambda x ,y: x if x > y else y,nums)
print(maximum)

sum=reduce(lambda acc , x: acc+x,nums)
print(sum)
```
