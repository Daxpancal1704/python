```python
class PowTwo:
    """class   to implement an iterator of powers of two"""
    def __init__(self,max=0):
        self.max=max

    def __iter__(self):
        self.n=0
        return self

    def  __next__(self):
        if self.n<=self.max:
            result=2**self.n
            self.n+=1
            return result
        else:
            raise StopIteration
num=PowTwo(3)

i=iter(num)

print(next(i))
print(next(i))
```

```python
def function(path):
    try:
        with open(path)as file:
            return file.read()
    except FileNotFoundError as e:
        print("Error: ",e)

try:
    function("hello.txt")
    print(function("hello.txt"))
except FileNotFoundError as e:
    print("error")     
```

```python
class InsufficientFundsError(Exception):
    """Raise when a withdrwal bigger than amount"""

def withdrwal(amount,balance):
        if amount < 0:
            raise ValueError ("Value must be positive")
        if amount > balance:
            raise InsufficientFundsError(f"balance is {balance} less then {amount}")
        return amount-balance
for amt in (100,-10,3000,):
    try:
        print("new balance : ", withdrwal(100,amt))
    except InsufficientFundsError as e:
         print("Balance is insuffucient",repr(e.__cause__))
    except ValueError as e:
         print("Bad amount",repr(e.__cause__))
```
