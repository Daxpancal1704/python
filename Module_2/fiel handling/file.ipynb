| **Mode** | **Meaning**              | **Creates File if Missing?** | **Errors if File Missing?** |      **Overwrites Existing Content?**      |
| :------: | ------------------------ | :--------------------------: | :-------------------------: | :----------------------------------------: |
|   `'r'`  | Read (text)              |              No             |  Yes (`FileNotFoundError`) |                     No                    |
|   `'w'`  | Write (text)             |              Yes            |              No            |        Yes (clears existing content)      |
|   `'a'`  | Append (text)            |              Yes            |              No            |           No (writes at the end)          |
|   `'x'`  | Exclusive create         |              Yes            |              No            |      No (fails if file already exists)    |
|  `'r+'`  | Read and write           |              No             |  Yes (`FileNotFoundError`) |  No (writes from current cursor position) |
|  `'w+'`  | Write and read           |              Yes            |              No            |        Yes (clears existing content)      |
|  `'a+'`  | Append and read          |              Yes            |              No            |           No (appends at the end)         |
