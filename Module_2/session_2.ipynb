### List Comprehension
A list comprehension is a concise, single-line syntax in Python for creating a new list by applying an expression to each item in an iterable, optionally filtering items with a condition.


A list comprehension is just a shorter, cleaner way to write a for loop that creates a list.

Basic syntax:

[expression for item in iterable]

[expression for item in iterable if condition]



````python
# Using a for loop
numbers = []
for i in range(5):
    numbers.append(i * 2)
print(numbers)
````

````python
numbers = [i * 2 for i in range(5)]
print(numbers)
````

##### More basic examples - transforming and filtering items

````python
words = ["python", "java", "rust"]

upper_words = [w.upper() for w in words]     # transform each item
lengths     = [len(w) for w in words]        # call a function on each item
first_chars = [w[0] for w in words]          # index into each item
letters     = [ch for ch in "hello"]         # any iterable works - even a string

print(upper_words)
print(lengths)
print(first_chars)
print(letters)
````

````python
names = ["Asha", "Ravi", "Anil", "Meera", "Arjun"]

# filter: keep only the names that start with "A"
a_names = [n for n in names if n.startswith("A")]
print(a_names)

# comprehensions can call your own functions too
def square(x):
    return x * x

print([square(n) for n in range(1, 6)])
````

````python
evens = [n for n in range(20) if n % 2 == 0]
evens
````

````python
#conditional list comprehension if/else
# [expr_if_true if condition else expr_if_false for item in iterable]

labels = ["even" if n % 2 == 0 else "odd" for n in range(10)]
labels
````

##### Position of `if` matters

| Where the `if` is | Name | What it does |
| --- | --- | --- |
| at the **end** `[x for x in data if cond]` | **filter** | items failing `cond` are **dropped** (output can be shorter) |
| at the **front** `[a if cond else b for x in data]` | **transform** | **every** item stays, only its value changes (same length) |

````python
nums = range(10)

filtered = [n for n in nums if n % 2 == 0]            # FILTER
print(filtered, len(filtered))

transformed = [n if n % 2 == 0 else -n for n in nums]  # TRANSFORM
print(transformed, len(transformed))

# both together: first filter (n > 2), then transform what is left
both = [n * 10 if n % 3 == 0 else n for n in nums if n > 2]
print(both)
````

##### Two shapes of nested comprehension

* **One flat list** -> two `for` clauses in one comprehension. They read left to right, exactly like nested loops written top to bottom.
* **A list of lists (matrix)** -> a comprehension *inside* a comprehension. The inner one builds each row.

````python
# one flat list
flat_pairs = [(r, c) for r in range(2) for c in range(3)]
print(flat_pairs)

# list of lists: the inner [...] builds one row
grid = [[r * 3 + c for c in range(3)] for r in range(3)]
print(grid)

# classic: transpose a matrix (rows become columns)
matrix = [[1, 2, 3],
          [4, 5, 6]]
transposed = [[row[i] for row in matrix] for i in range(3)]
print(transposed)
````

##### When NOT to use a comprehension
* More than two `for` clauses, or a long condition -> use a normal loop (readability wins).
* You only want a **side effect** (printing, writing to a file, appending elsewhere) -> use a loop; a comprehension is for **building a value**.
* The data is huge and you only need to loop over it once -> use a **generator expression** `( ... )` (covered below).

````python
# enumerate() is a built-in Python function that lets you loop through a collection while keeping track of the index.

fruits = ["apple", "banana", "orange"]

index = 0
for fruit in fruits:
    print(index, fruit)
    index += 1

for i , fruit in enumerate(fruits):
    print(i,fruit)
````

````python
# zip() is a built-in Python function that combines two or more iterables (like lists or tuples) element by element.
names = [0, 1, 2,3]
scores = ['apple','banan','orange',4,5]

for name, score in zip(names, scores):
    print(name, score)
````

##### enumerate() and zip() - a few more tricks

````python
fruits = ["apple", "banana", "orange"]

# enumerate: choose where the counter starts
for rank, fruit in enumerate(fruits, start=1):
    print(rank, fruit)
print()
# zip with three iterables
names  = ["Asha", "Ravi", "Meera"]
marks  = [88, 92, 79]
cities = ["Pune", "Delhi", "Goa"]
for n, m, c in zip(names, marks, cities):
    print(f"{n:<6} {m:>3}  {c}")
print()
# zip + dict() -> quickest way to build a dictionary from two lists
print(dict(zip(names, marks)))

# "unzip" with *
pairs = [("a", 1), ("b", 2), ("c", 3)]
letters, numbers = zip(*pairs)
print(letters, numbers)
````

````python
# zip stops at the SHORTEST iterable - extra items are silently dropped
print(list(zip([1, 2, 3], ["x", "y"])))

# zip_longest keeps everything and fills the gaps
from itertools import zip_longest
print(list(zip_longest([1, 2, 3], ["x", "y"], fillvalue="-")))

# Python 3.10+: strict=True raises an error if lengths differ
try:
    list(zip([1, 2, 3], ["x", "y"], strict=True))
except ValueError as e:
    print("ValueError:", e)

# enumerate + zip + comprehension together
print([f"{i}:{n}" for i, (n, m) in enumerate(zip(names, marks)) if m > 80])
````

##### `enumerate()` vs `zip()`

| | `enumerate(iterable)` | `zip(a, b, ...)` |
| --- | --- | --- |
| **Purpose** | adds a **counter** (index) to **one** iterable | pairs up items from **two or more** iterables |
| **Yields** | `(index, item)` | `(item_from_a, item_from_b, ...)` |
| **Stops when** | the iterable ends | the **shortest** iterable ends |
| **Typical use** | `for i, x in enumerate(items, start=1)` | `dict(zip(keys, values))` |

##### Other comprehensions in Python

| **Comprehension Type**       | **Syntax**                    | **Returns**                                   | **Example Output**       |
| ---------------------------- | ----------------------------- | --------------------------------------------- | ------------------------ |
| **List Comprehension**       | `[x for x in iterable]`       | `list`                                        | `[1, 2, 3]`              |
| **Set Comprehension**        | `{x for x in iterable}`       | `set` (removes duplicates)                    | `{1, 2, 3}`              |
| **Dictionary Comprehension** | `{k: v for k, v in iterable}` | `dict`                                        | `{'a': 1, 'b': 2}`       |
| **Generator Expression**     | `(x for x in iterable)`       | `generator` (lazy, produces values on demand) | `<generator object ...>` |

````python
names = ["Asha", "Ravi", "Meera"]
marks = [88, 92, 79]

score_book = {n: m for n, m in zip(names, marks)}              # from two lists
squares    = {n: n ** 2 for n in range(1, 6)}                  # computed values
passed     = {n: m for n, m in score_book.items() if m >= 80}  # filter a dict
curved     = {n: m + 5 for n, m in score_book.items()}         # transform values
inverted   = {m: n for n, m in score_book.items()}             # swap keys and values
word_len   = {w: len(w) for w in ["python", "is", "fun"]}

for d in (score_book, squares, passed, curved, inverted, word_len):
    print(d)
````

**Rules to remember**
* In a dict comprehension, if the same key is produced twice, the **last value wins**.
* Keys of a dict and items of a set must be **hashable** (immutable): `int`, `str`, `tuple` are fine, `list`/`dict`/`set` are not -> `TypeError: unhashable type: 'list'`.
* Tuple has **no** comprehension: `(x for x in data)` is a **generator expression**, not a tuple. Use `tuple(x for x in data)`.

#### Generators
A generator is an object that produces values one at a time, only when you ask for them, instead of creating and storing all the values at once.

This behavior is called lazy evaluation.

##### List comprehension vs generator expression

| | List comprehension `[...]` | Generator expression `(...)` |
| --- | --- | --- |
| Result | a **list** with all the values | a **generator** that produces values on demand |
| Memory | stores every item at once | stores almost nothing |
| Reusable | yes (index, `len`, loop many times) | **no** - one pass only |
| Use when | you need the data several times | you loop once, or the data is huge |

````python
numbers = [x * 2 for x in range(5)]
print(numbers)
# The entire list is created and stored in memory immediately.
````

````python
#generator comprehension creates an iterator that generates items on-the-fly and does not store the entire list in memory. It uses parentheses instead of square brackets.  
numbers = (x * 2 for x in range(5))
print(list(numbers))
````

````python
a=(i for i in range(10,4,-1))
sorted(a)
````

**Why do we see `<generator object ...>` instead of values?**
A generator does **not** compute anything when it is created - it only remembers *how* to produce the values. The values are calculated one by one when something asks for them: `next(gen)`, a `for` loop, `list(gen)`, `sum(gen)` and so on. Printing the generator itself just shows the object.

````python
numbers = (x * 2 for x in range(5))

print(next(numbers))
print(next(numbers))
print(next(numbers))
print(next(numbers))
print(next(numbers))
try:
    print(next(numbers))        # nothing left -> StopIteration
except StopIteration:
    print("StopIteration: the generator is exhausted")
````

##### Iterable vs Iterator

| | **Iterable** | **Iterator** |
| --- | --- | --- |
| Meaning | something you can loop over | an object that hands out items **one at a time** |
| Has | `__iter__()` | `__iter__()` **and** `__next__()` |
| Examples | list, tuple, str, dict, set, `range` | `iter(list)`, file objects, **generators** |
| Reusable? | yes - every `for` loop starts fresh | **no** - once exhausted it stays empty |

````python
nums = [10, 20, 30]      # a list is ITERABLE
it = iter(nums)          # iter() gives its ITERATOR
print(next(it), next(it), next(it))
# next(it)               # -> StopIteration: nothing left

# next() accepts a default so you never hit StopIteration
print(next(it, "no more items"))

# this is what a for loop does behind the scenes:
it = iter(nums)
while True:
    try:
        item = next(it)
    except StopIteration:
        break
    print(item)
````

##### Why use generators?
1. **Memory** - a generator holds one item at a time, a list holds all of them.
2. **Speed to first result** - you can start using values immediately instead of waiting for the whole list to be built.
3. **Infinite or very long sequences** - possible only because values are produced lazily.
4. **Pipelines** - chain steps (read -> clean -> filter) without creating big temporary lists.
5. **Large files** - process a file line by line.

````python
# Two Ways to Create a Generator

# 1.
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

gen = count_up_to(5)
print(gen)   



# yield is the key — it's like return, but instead of ending the function, it pauses it and hands back a value. Next time you ask for a value, execution resumes right after the yield.

````

````python
# yield PAUSES the function and remembers where it stopped
def demo():
    print("start")
    yield 1
    print("resumed after 1")
    yield 2
    print("resumed after 2")
    yield 3
    print("end")

g = demo()          # nothing is printed yet - the body has not started running
print(next(g))
print(next(g))
print(next(g))
# next(g)           # function finishes -> StopIteration
````

| `return` | `yield` |
| --- | --- |
| ends the function for good | **pauses** the function, keeps local variables |
| gives back **one** value | can give back **many** values over time |
| function runs again from the top | next `next()` resumes right after the `yield` |

````python
# Four ways to consume a generator
print(list(count_up_to(5)))            # 1. list()  - takes everything

g = count_up_to(5)
print(next(g), next(g))                # 2. next()  - one item at a time

for x in count_up_to(3):               # 3. for loop - handles StopIteration for you
    print(x)

print(sum(count_up_to(4)), max(count_up_to(4)), tuple(count_up_to(3)))   # 4. other consumers: sum, max, tuple, set, sorted ...
````

##### Generators - advanced

````python
# An INFINITE sequence - only possible because generators are lazy
from itertools import islice

def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

print(list(islice(fibonacci(), 10)))     # take only the first 10
````

````python
# yield from -> hand over to another generator / iterable
def flatten(nested):
    for item in nested:
        if isinstance(item, list):
            yield from flatten(item)      # recursion inside a generator
        else:
            yield item

print(list(flatten([1, [2, [3, 4]], 5])))
````

````python
# Generator PIPELINE: each stage pulls one item at a time from the previous one.
# At no point does a full list exist in memory.
lines   = ["1", "2", "3", "4", "5", "6"]
numbers = (int(x) for x in lines)
evens   = (n for n in numbers if n % 2 == 0)
squares = (n * n for n in evens)

print(sum(squares))      # 4 + 16 + 36 = 56
````

````python
# 'r+' -> read AND write; the cursor starts at 0 and write() overwrites in place
with open("sample.txt", "w") as f:
    f.write("Hello World\nsecond line\nthird line\n")

with open("sample.txt", "r+") as f:
    f.write("HELLO")           # overwrites the first 5 characters
    f.seek(0)
    print(f.read())

# 'a+' -> append AND read; the cursor starts at the END, so seek(0) before reading
with open("sample.txt", "a+") as f:
    f.write("appended line\n")
    f.seek(0)
    print(f.read())
````

````python
# Reading a big file in fixed-size chunks. It is a generator - ties back to the last topic.
def read_in_chunks(path, size=10):
    with open(path, "r", encoding="utf-8") as f:
        while True:
            chunk = f.read(size)
            if not chunk:            # empty string means end of file
                break
            yield chunk

chunks = list(read_in_chunks("sample.txt", 10))
print(len(chunks), "chunks; first two:", chunks[:2])
print("total characters:", sum(len(c) for c in read_in_chunks("sample.txt")))
````

##### File paths: absolute vs relative, and the backslash trap
* **Relative path** `data/file.txt` -> resolved from the current working directory (`os.getcwd()`).
* **Absolute path** `C:/Users/me/data/file.txt` -> full location, works from anywhere.
* On Windows `\` starts an **escape sequence** (`\n` = newline, `\t` = tab ...), so `"C:\new\test.txt"` silently breaks. Use a **raw string** `r"..."`, forward slashes, or `os.path.join` / `pathlib`.

````python
import os
print(os.getcwd())

print("C:\new_folder\test.txt")                  # broken: \n and \t became whitespace
print(r"C:\new_folder\test.txt")                 # raw string keeps the backslashes
print("C:/new_folder/test.txt")                   # forward slashes work on Windows too
print(os.path.join("C:\\", "new_folder", "test.txt"))   # builds the path for the current OS
````

### Error Handling

Error handling (exception handling) is the mechanism Python provides to detect, respond to, and recover from runtime errors — without crashing the entire program. Instead of letting an unexpected condition (missing file, bad input, division by zero) kill your script, you catch it and decide what to do next.

Python implements this through exceptions — objects raised when something goes wrong, which can be caught and handled using try/except blocks.

exceptions are for exceptional, unexpected situations — not a substitute for a simple check that already has a clean non-exception solution



##### Three kinds of errors

| Type | When it shows up | Can `try/except` catch it? |
| --- | --- | --- |
| **Syntax error** | before the program runs (bad grammar) | No - fix the code |
| **Runtime error (exception)** | while the program runs (bad input, missing file ...) | **Yes** |
| **Logical error** | never - program runs but gives a **wrong answer** | No - find it by testing |

````python
# 1. SyntaxError - Python refuses to even start
# print("hello"            <- missing bracket

# 2. Runtime error - raised while running; can be handled
# 10 / 0                   <- ZeroDivisionError

# 3. Logical error - no message at all, just a wrong answer
def average(a, b):
    return a + b / 2       # bug: should be (a + b) / 2

print(average(10, 20))     # prints 20.0, the right answer is 15.0
````

````python
# full block structure: try / except / else / finally
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("division error")
    except Exception as e:
        print(f"Unexpected error: {e}")
    else:
        print("Ran because NO exception occurred:", result)
    finally:
        print("Always runs - exception or not")
        print("-" * 30)

divide(10, 2)
divide(10, 0)
divide(10, "a")

# Why else exists: code that should only run on success, but that you don't want inside the try block
# (so it isn't accidentally caught by your own except if it also raises something similar).
# It keeps the try block minimal - only the actual risky line is inside it.
````

````python
# trigger each one safely. (lambda = a tiny one-line function)
def trigger(label, func):
    try:
        func()
    except Exception as e:
        print(f"{label:<18} -> {type(e).__name__}: {e}")

trigger("int('abc')",      lambda: int("abc"))
trigger("'a' + 1",         lambda: "a" + 1)
trigger("{}['x']",         lambda: {}["x"])
trigger("[1, 2][5]",       lambda: [1, 2][5])
trigger("1 / 0",           lambda: 1 / 0)
trigger("open('nofile')",  lambda: open("nofile.txt"))
trigger("None.upper()",    lambda: None.upper())
trigger("undefined_name",  lambda: undefined_name)  
trigger("import nothing",  lambda: __import__("nothing_here"))
trigger("next(iter([]))",  lambda: next(iter([])))
````

````python
# Several exceptions, same handling -> put them in a tuple
def parse_and_divide(text, divisor):
    try:
        return int(text) / divisor
    except (ValueError, TypeError):
        print("  bad input")
    except ZeroDivisionError:
        print("  divisor is zero")

for text, d in [("10", 2), ("x", 2), ("10", 0), (None, 2)]:
    print(text, d, "->", parse_and_divide(text, d))
````

````python
# ORDER matters: Python uses the FIRST matching except block.
# Put specific exceptions first and general ones last.
try:
    [1, 2, 3][10]
except LookupError:           # parent class - catches IndexError AND KeyError
    print("lookup problem")
except IndexError:            # never reached: the parent already caught it
    print("never printed")

# Avoid a bare "except:" - it also swallows Ctrl+C and hides real bugs.
# Prefer "except Exception:" at the very least, and a specific class whenever you can.
````

````python
# Re-raise: log something, then pass the SAME exception to the caller
def read_config(path):
    try:
        with open(path) as f:
            return f.read()
    except FileNotFoundError:
        print(f"[log] config file missing: {path}")
        raise                               # a bare raise re-raises it unchanged

try:
    read_config("missing.cfg")
except FileNotFoundError:
    print("caller handled it")
````

````python
# Custom exceptions: an (almost) empty class that inherits from Exception
# (classes are a later topic - for now treat this as a template)
class InsufficientFundsError(Exception):
    """Raised when a withdrawal is bigger than the balance."""
    pass

def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("amount must be positive")
    if amount > balance:
        raise InsufficientFundsError(f"balance {balance} is less than {amount}")
    return balance - amount

for amt in (30, 500, -5):
    try:
        print("new balance:", withdraw(100, amt))
    except InsufficientFundsError as e:
        print("Funds problem:", e)
    except ValueError as e:
        print("Bad amount:", e)
````

````python
# Exception chaining: raise a clearer error but keep the original cause
def load_age(text):
    try:
        return int(text)
    except ValueError as e:
        raise RuntimeError(f"Could not load age from {text!r}") from e

try:
    load_age("abc")
except RuntimeError as err:
    print(err)
    print("caused by:", repr(err.__cause__))
````

````python
# Retry pattern: keep asking until the input is valid
def ask_int(prompt, low, high):
    while True:
        try:
            n = int(input(prompt))
        except ValueError:
            print("Please type digits only.")
            continue
        if low <= n <= high:
            return n
        print(f"Number must be between {low} and {high}.")

# age = ask_int("Age (1-120): ", 1, 120)     # uncomment to try it
````

````python
# A realistic helper: file + JSON + several failure modes
import json

def safe_load_json(path, default=None):
    try:
        with open(path, "r", encoding="utf-8") as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"{path} not found - using default")
    except json.JSONDecodeError as e:
        print(f"{path} is not valid JSON (line {e.lineno}): {e.msg}")
    except PermissionError:
        print(f"no permission to read {path}")
    return default

with open("bad.json", "w") as f:
    f.write("{ 'name': 'Alice' ")           # invalid JSON on purpose

print(safe_load_json("does_not_exist.json", default={}))
print(safe_load_json("bad.json", default={}))
````

##### EAFP vs LBYL - and when NOT to use exceptions
* **LBYL** - *Look Before You Leap*: check first with `if`.
* **EAFP** - *Easier to Ask Forgiveness than Permission*: just try, and handle the failure. This is the Pythonic style.
* Best of all: use the tool built for the job (like `dict.get()` above) when one exists.

````python
data = {"name": "Asha"}

# LBYL
if "age" in data:
    age = data["age"]
else:
    age = None

# EAFP
try:
    age = data["age"]
except KeyError:
    age = None

# built-in shortcut
age = data.get("age")
print(age)
````

##### Error handling - good habits
1. Catch **specific** exceptions, not everything.
2. Keep the `try` block **small** - only the risky line(s).
3. Never silence errors with `except: pass`; at least log them.
4. Use `finally` / `with` for cleanup (closing files, connections).
5. Raise errors with a **meaningful message**, using the most fitting exception type.
6. Do not use exceptions for normal flow when a simple `if` is clearer.
