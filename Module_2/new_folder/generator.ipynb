```python
# Two Ways to Create a Generator

# 1.

def count(n):
    i=1
    while i <=n:
        yield i
        i+=1
print(next(count(10)))

# yield is the key — it's like return, but instead of ending the function, it pauses it and hands back a value. Next time you ask for a value, execution resumes right after the yield.

```

```python
# Two Ways to Create a Generator

# 1.
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

gen = count_up_to(5)
print(next(gen))   
print(next(gen))   
print(next(gen))   



# yield is the key — it's like return, but instead of ending the function, it pauses it and hands back a value. Next time you ask for a value, execution resumes right after the yield.

```

```python
"Generator advanced"
# An INFINITE sequence - only possible because generators are lazy
# from itertools import islice

# def fibonacci():
#     a, b = 0, 1
#     while True:
#         yield a
#         a, b = b, a + b

# print(list(islice(fibonacci(), 10)))     # take only the first 10


from itertools import islice

def febo():
    a,b=0,1
    while True:
        yield a
        a,b=b,a+b
list(islice(febo(),10))        
```

```python
def funct(nested):
    for item in nested:
        if isinstance(item,list):
            yield from funct(item)
        else:
            yield item


print(list(funct([1, [2, [3, 4]], 5])))
```
