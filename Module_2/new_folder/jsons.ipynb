```python
#if write in json file so yu can use json.dumps else you can only read so use json.loads

import json
var = { 
      "Subjects": {
                  "Maths":85,
                  "Physics":90,
                  "Science":90
                   }
      }

with open("Sample.json", "w") as p:
     json.dump(var, p,indent=4)
with open("Sample.json", "r") as read_it:
     data = json.load(read_it)
     print(data)
```
