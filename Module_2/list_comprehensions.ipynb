```python
fruits=["apple","banana","orange","mango"]

upper=[f.upper() for f in fruits]
capital=[c.capitalize() for c in fruits]
find_char=[f[1] for f in fruits]
first_le=[l[0] for  l in fruits]
print(upper)
print(capital)
print(find_char)
print(first_le)
```

```python
"Even or odd"

even=[x for x in range(100) if x % 2== 0 ]
print(even)
#conditional list comprehension if/else
# [expr_if_true if condition else expr_if_false for item in iterable]

check_num=["odd" if i % 2 ==0 else "odd" for i in range(1,100,5) ]
print(check_num)
name="Dax panchal hello "
vowels="aeiou"


check_vowels=[ch for ch in name.lower() if ch in vowels]
print(len(check_vowels))



```

```python
num=range(10)
# filter
check=[n for n in num if n %2 == 0]
print(check , len(check))

#transform
chec=[n if n %2 != 0 else -n for n in num]
checks=[n if n%2==0 else -n for n in num]
print(checks)
print(chec)



# both together: first filter (n > 2), then transform what is left
# both = [n * 10 if n % 3 == 0 else n for n in nums if n > 2]
# print(both)


both=[n *2 if n %2!=0 else n for n in num if n >2]
print(both)
```

```python
"Nested loopa in comprehnions"

loop=[(i,j) for i in range(3) for j in range(3)]
print(len(loop))
```

```python
"Nested loop with condtions"


loops=[(i,j) for i in range(1,11) for j in range(1,11)  if i != j ]
# len(loop)
loops=[(a,b) for a in  range(21) for b in range(21) if a != b]
print(len(loops))
print(loops)
print(loops[:10])
```

```python
"enumerate fucntion to use kep track the index"

f_name=["alice","Bob"]
l_name=["john","den"]

for i ,(f_name ,l_name) in enumerate(zip(f_name,l_name)):
    print(i,f_name,l_name)
```
