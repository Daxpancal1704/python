Class variable, classmethod and staticmethod 
* A class vairable count() that increae by 1 every time a student object is created.
* A class method get_count() and return count.
* A static method is_pass(marks) that return true when marks are 50 or more 

```python
class student:

    count=0

    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
        student.count +=1

    @classmethod
    def get_count(cls):
        return cls.count

    @staticmethod
    def is_pass(marks):
        return marks >= 50 

s1=student("alice",90)
s2=student("alice",20)
s3=student("alice",20)

#Methods always calling to class name but ike a static method ,classmethods

print(student.get_count())
print(student.is_pass(s1.marks),student.is_pass(s2.marks))
```
