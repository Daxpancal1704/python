# Q4 . functional programming
* Read one line of space-separated integer.Using lambda with map(), filter() and reduce()(without no loops and no comprehensions)
print:
1. the cube of every number(as a list)
2. only the numbers divisible by 3(as a list)
3. the product of al the number

```python

from functools import reduce

numbers = list(map(int, input("Enter numbers: ").split()))

# 1. Cube of every number
cube = list(map(lambda x: x ** 3, numbers))
print(cube)

# 2. Numbers divisible by 3
divisible = list(filter(lambda x: x % 3 == 0, numbers))
print(divisible)

# 3. Product of all numbers
product = reduce(lambda x, y: x * y, numbers)
print(product)
```
